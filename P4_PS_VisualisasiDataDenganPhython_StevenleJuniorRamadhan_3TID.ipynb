```python
import pandas as pd
import matplotlib.pyplot as plt

path = "StevenleJuniorRamadhan_3TID.xlsx"
dataraw = pd.read_excel(path, header=7)
dataraw = dataraw.iloc[:, 1:8].copy()
dataraw = dataraw.dropna(how='all', axis=1)
dataraw.columns = dataraw.columns.astype(str).str.strip()
dataraw = dataraw.dropna(subset=['Grade']).copy()
dataraw.columns = dataraw.columns.str.strip()

print("--- DATA MENTAH ---")
print(dataraw.columns.tolist())
print(dataraw.head())

datafrq = pd.crosstab(index=dataraw["Grade"], columns="Frekuensi")
print("\n--- Tabel Frekuensi ---")
print(datafrq)
```

```python
plt.figure(figsize=(6, 4))
plt.plot(datafrq.index, datafrq['Frekuensi'], marker='o', color='b', label='Frekuensi')
plt.title('Grafik Garis Distribusi Grade')
plt.xlabel('Grade')
plt.ylabel('Frekuensi')
plt.plot(datafrq.index, datafrq['Frekuensi'], marker='o', color='purple')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6, 4))
datafrq['Frekuensi'].plot(kind='bar', color='purple')
plt.title('Grafik Batang Distribusi Grade')
plt.xlabel('Grade')
plt.ylabel('Frekuensi')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6, 6))
datafrq['Frekuensi'].plot(
    kind='pie',
    autopct='%1.0f%%',
    startangle=90,
    legend=False
)
plt.title('Pie Chart Persentase Grade')
plt.ylabel('')
plt.tight_layout()
plt.show()
```

```python
dataraw["Final Score"] = pd.to_numeric(dataraw["Final Score"], errors='coerce')
dt = dataraw["Final Score"]

stats = dt.describe()
stats['Standard Error'] = dt.sem()
stats['Median'] = dt.median()
stats['Mode'] = dt.mode().iloc[0]
stats['variance'] = dt.var()
stats['range'] = dt.max() - dt.min()
stats['skewness'] = dt.skew()
stats['kurtosis'] = dt.kurtosis()

print("\n--- HASIL STATISTIKA DESKRIPTIF ---")
print(stats)
```
